```python
import json
from collections import Counter

with open("metadata_jugadores.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# Soporte para lista o diccionario raíz
jugadores = data if isinstance(data, list) else list(data.values())

ligas = [j["info"]["liga"] for j in jugadores if j.get("info", {}).get("liga")]

unicas = sorted(set(ligas))
conteo = Counter(ligas)

print(f"Ligas únicas encontradas ({len(unicas)}):")
for liga in unicas:
    print(f"  - {liga}: {conteo[liga]} jugador(es)")
```

```python
import json
from collections import Counter

# Ruta al archivo JSON
ruta_json = "metadata_jugadores.json"

# Cargar datos
with open(ruta_json, "r", encoding="utf-8") as f:
    datos = json.load(f)

# Contar ligas
contador_ligas = Counter()

for jugador in datos:
    liga = jugador.get("info", {}).get("liga")
    
    if liga:
        contador_ligas[liga] += 1

# Mostrar resultados
print(f"Número de ligas diferentes: {len(contador_ligas)}\n")

for liga, cantidad in sorted(contador_ligas.items()):
    print(f"{liga}: {cantidad} jugadores")
```
